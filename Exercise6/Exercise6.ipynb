```python
import pandas as pd
import seaborn as sns

from matplotlib import pyplot as plt
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

```python
train_data = 'https://github.com/robitussin/CCMACLRL_EXERCISES/blob/3fd7d51ffd17863598ac3f44eeefc558171a5b73/dataset/house-prices-advanced-regression-techniques/train.csv?raw=true'
df = pd.read_csv(train_data)
```

###Data Preprocessing

####Dropping ID Column

```python
df = df.drop('Id', axis=1)
```

```python
df.head()
```

#####Check for missing data

```python
df.isnull().sum()
```

####Identify numerical and categorical columns

```python
df_num_cols = df.select_dtypes(include=['int64', 'float64']).columns.drop('SalePrice')
df_cat_cols = df.select_dtypes(include=['object']).columns
```

##### Impute missing numerical values with median

```python
from sklearn.impute import SimpleImputer
df_num_imputer = SimpleImputer(strategy='median')
df[df_num_cols] = df_num_imputer.fit_transform(df[df_num_cols])
```

##### Impute categorical missing values with 'None'


```python
df_cat_imputer = SimpleImputer(strategy='constant', fill_value='None')
df[df_cat_cols] = df_cat_imputer.fit_transform(df[df_cat_cols])
```

```python
# One-hot encode categorical variables
df = pd.get_dummies(df, columns=df_cat_cols, drop_first=True)
```

```python
df.info()
```

```python
df
```

###Split Data

```python
from sklearn.model_selection import train_test_split
X = df.drop("SalePrice", axis=1)
y = df["SalePrice"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
```

## Sample Submission File

```python
sample_submission_url ='https://github.com/robitussin/CCMACLRL_EXERCISES/blob/3fd7d51ffd17863598ac3f44eeefc558171a5b73/dataset/house-prices-advanced-regression-techniques/sample_submission.csv?raw=true'

sf=pd.read_csv(sample_submission_url)
```

## 1. Train a KNN Regressor

```python
score_list = {} # I've created this dict for saving score variables into it
```

```python
from sklearn.neighbors import KNeighborsRegressor
KNN = KNeighborsRegressor(n_neighbors=22) #I've tried more than 50 values. 22 is the best value

KNN.fit(X_train,y_train)
knn_score = KNN.score(X_test,y_test)
score_list["KNN Regression"] = knn_score
print(f"Score is {knn_score}")
```

- Perform cross validation

```python
from sklearn.model_selection import cross_val_score
#five folds cross validation
knn_scores = cross_val_score(KNN, X, y, cv=5)

print(f"KNN Regressor cross-validation scores: {knn_scores}")
print(f"Average KNN Regressor cross-validation score: {knn_scores.mean()}")
```

## 2. Train a SVM Regression

```python
from sklearn.svm import SVR

svr = SVR()
svr.fit(X, y)

score_list['SVR'] = svr.score(X, y)
print(f"SVM Regressor score: {score_list['SVR']}")
```

- Perform cross validation

```python
from sklearn.model_selection import cross_val_score
#five folds cross validation
svr_scores = cross_val_score(svr, X, y, cv=5)

print(f"SVR Regressor cross-validation scores: {svr_scores}")
print(f"Average SVR Regressor cross-validation score: {svr_scores.mean()}")
```

## 3. Train a Decision Tree Regression

```python
from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor(random_state=1)
dt.fit(X, y)

score_list['Decision Tree'] = dt.score(X, y)
print(f"Decision Tree Regressor score: {score_list['Decision Tree']}")
```

- Perform cross validation

```python
from sklearn.model_selection import cross_val_score
#five folds cross validation
dt_scores = cross_val_score(dt, X, y, cv=5)

print(f"Decision Tree Regressor cross-validation scores: {dt_scores}")
print(f"Average Decision Tree Regressor cross-validation score: {dt_scores.mean()}")
```

## 4. Train a Random Forest Regression

```python
from sklearn.ensemble import RandomForestRegressor

rfr = RandomForestRegressor(n_estimators=50, random_state=1)
rfr.fit(X, y)

score_list['Random Forest'] = rfr.score(X, y)
print(f"Random Forest Regressor score: {score_list['Random Forest']}")
```

```python
from sklearn.model_selection import cross_val_score
#five folds cross validation
rfr_scores = cross_val_score(rfr, X, y, cv=25)

print(f"Random Forest Regressor cross-validation scores: {rfr_scores}")
print(f"Average Random Forest Regressor cross-validation score: {rfr_scores.mean()}")
```

## 5. Compare all the performance of all regression models

```python
print("\nCross-validation Scores:")
print(f"KNN Regressor: {knn_scores.mean()}")
print(f"SVR Regressor: {svr_scores.mean()}")
print(f"Decision Tree Regressor: {dt_scores.mean()}")
print(f"Random Forest Regressor: {rfr_scores.mean()}")
```
